# Steamming
## comments of  product is a positive review or negative review
## review --> eating, eat, eaten [going , gone, goes ] --> go


```python
words = [ "eating","eat","eaten","write","written","writes","program","programming","Programs","history","finally","finalized"]
```

## PorterStemmer ( a type of streamming)

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"--->" + stemming.stem(word))
```

## RegexpStemmer class

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|ables$', min=4)
```

```python
reg_stemmer.stem("Running")
```

```python
reg_stemmer.stem("ingRRunning")
```

## SnowBall stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowballsstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"===>" + snowballsstemmer.stem(word))
```

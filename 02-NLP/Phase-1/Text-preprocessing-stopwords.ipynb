## Text-preprocessing-stopwords
Stop Words in Text Preprocessing

Stop words are common words that appear very frequently in a language but, in many NLP tasks, carry relatively little useful information

```python
### speach on Quantum physics by gpt
paragraph ="""Quantum physics is the branch of physics that studies the behavior of very small particles, such as atoms, electrons, and photons. At this tiny scale, nature behaves very differently from what we experience in our everyday lives.
One of the most interesting ideas in quantum physics is quantum superposition. It means that a particle can exist in multiple possible states until it is measured. Another important concept is quantum entanglement, where two particles can become connected in such a way that the state of one is related to the state of the other, even when they are very far apart.
Quantum physics also tells us that we cannot always predict exactly what will happen to a particle. Instead, we can calculate the probability of different outcomes. This changed our traditional understanding of how the universe works.
Quantum physics is not just a theoretical subject. It has already given us many technologies that we use today, including lasers, computer chips, LEDs, medical imaging technologies, and modern communication systems. Today, scientists are also working on quantum computers, which could solve certain types of problems much faster than conventional computers.
What makes quantum physics truly exciting is that it challenges our common sense. The rules that seem strange at the atomic level are actually fundamental to how our universe works.
To conclude, quantum physics teaches us an important lesson: the universe is much more mysterious and interesting than what we can see with our eyes.
"""
```

```python
from nltk.stem import PorterStemmer
```

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
sentences =nltk.sent_tokenize(paragraph)
```

```python
type(sentences)
```

```python
# Apply stopwords and filter and then Apply Stemming
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)  ## converting all the list words into sentences
```

```python
sentences
```

```python
from nltk.stem import SnowballStemmer
snowballstemmer = SnowballStemmer('english')
```

```python
# Apply stopwords and filter and then Apply Snowball Stemming
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[snowballstemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)  ## converting all the list words into sentences
```

```python
sentences
```

```python
## now with the help of lemmatization
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
# Apply stopwords and filter and then Apply lemmatization Stemming
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[ lemmatizer.lemmatize(word, pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)  ## converting all the list words into sentences
```

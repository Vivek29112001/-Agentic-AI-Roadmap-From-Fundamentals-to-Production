 # wordnet  Lemmatizer

 ### The WordNetLemmatizer is a popular Python tool in the NLTK Library that reduces inflected words to their correct base or dictionary form, known as a lemma.
 #### How It Works
 It uses WordNet, a large lexical database of English, to look up base words.It defaults to treating input words as nouns ('n') if you do not specify a part-of-speech tag.It accepts parameters like 'n' for noun, 'v' for verb, 'a' for adjective, and 'r' for adverb

```python
import nltk

nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("going",pos='n')
```

```python
words = [ "eating","eat","eaten","write","written","writes","program","programming","Programs","history","finally","finalized"]
```

```python
for word in words:
    print(word+"===>" + lemmatizer.lemmatize(word, pos='v'))
```

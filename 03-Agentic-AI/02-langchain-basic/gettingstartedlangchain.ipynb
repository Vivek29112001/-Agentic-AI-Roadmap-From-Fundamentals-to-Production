```python
## prompt engineering

from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "you are an expert AI engineer. provide me answer baed on the questions"),
        ("user", "{input}")
    ]
)
prompt
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-20b")
model
```

```python
### Chaining -> combing prompt with LLM

chain = prompt | model
chain
```

```python
response = chain.invoke({"input":"Can u tell me about langsmith"})
print(response.content)
```

```python
## outerputparser -> way to display the output

from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

chain = prompt | model | output_parser

response = chain.invoke({"input":"Can u tell me about langsmith"})
print(response)
```

```python
### outputparser


from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import PromptTemplate


output_parser = JsonOutputParser()

prompt = PromptTemplate(
    template="Answer tghe user query \n {format_instruction}\n {query}\n",
    input_variables=["query"],
    partial_variables={"format_instruction":output_parser.get_format_instructions()},
)
```

```python
# Task what we have done with Prompt template now lerts try with chat prompt template

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

template = ChatPromptTemplate(
    {
        ("system", "You are a helpful AI bot. Your name is {name}."),
        ("user", "{user_input}"),
    }
)
```

```python
## web based loader
from langchain_community.document_loaders import WebBaseLoader
import bs4
loader = WebBaseLoader(web_paths=("https://docs.aws.amazon.com/AWSEC2/latest/UserGuide/concepts.html","https://www.website.com/?source=SC&country=IN","https://docs.aws.amazon.com/decision-guides/latest/decision-guides/choosing-aws-serverless-service.html"),
                       bs_kwargs=dict(parse_only = bs4.SoupStrainer(
                           class_=("title","post-content","header","awsdocs-container")
                       )))

web_documents= loader.load()
web_documents
```

```python

import arxiv
from langchain_core.documents import Document

client = arxiv.Client()

search = arxiv.Search(
    id_list=["1706.03762"],
    max_results=2
)

results = list(client.results(search))

docs = [
    Document(
        page_content=f"{result.title}\n\n{result.summary}",
        metadata={
            "title": result.title,
            "authors": [author.name for author in result.authors],
            "published": str(result.published),
            "url": result.entry_id
        }
    )
    for result in results
]

docs
```

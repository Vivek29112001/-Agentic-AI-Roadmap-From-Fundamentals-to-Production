
## What are pydantic models?
#### Pydentic models are the foundation of data validatiopn in python . They use python type annotatios to define the structure and validation data at runtime.

```python
from dataclasses import dataclass

@dataclass
class Person():
    name:str
    age:str
    city:str
```

```python
person =Person(name="vivek", age=24, city="Nanital")
print(person)
```

```python
## pydantic is used to check the data type means check in every point

from pydantic import BaseModel
```

```python
## This is a Data model
class Person1(BaseModel):
    name:str
    age:int
    city:str
    
person =Person1(name="vivek", age=24, city="Nanital")
print(person)
```

```python
## models with optional fields

from typing import Optional

class Employee(BaseModel):
    id: int
    name:str
    department:str
    salary:Optional[float]= None  # optional with default value
    is_active: Optional[bool]=True # optional field with default value
```

```python
emp1= Employee(id=1,name="john",department="CSE")
print(emp1)
```

```python
emp2= Employee(id=1,name="john sike",department="CSE", salary="98000")
print(emp2)
```
